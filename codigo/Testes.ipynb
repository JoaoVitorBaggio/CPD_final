# Árvore de Prefixos
Inicializa

```python
from TST        import TST
from caminhos   import pasta_dados

arvore = TST(caminho=pasta_dados / "players.csv")
```

Pesquisa

```python
prefixo = "Maria"

palavras_com_ids = arvore.collect_words_with_prefix(prefixo)

if palavras_com_ids:
    for palavra, id_valor in palavras_com_ids:
            print(f'ID: {id_valor}, Nome: {palavra}')
else:
    print(f'Nenhuma palavra encontrada com o prefixo "{prefixo}".')
```
