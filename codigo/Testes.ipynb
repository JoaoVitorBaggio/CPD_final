### Árvore de Prefixos

Inicializa

```python
from TST        import TST
from caminhos   import pasta_dados
from jogadores  import ordenar_jogadores_por_medias

arvore = TST(caminho=pasta_dados / "players.csv")
```

Pesquisa

```python
prefixo = "B"

ids = arvore.collect_words_with_prefix(prefixo)
jogadores = [tabela_jogadores[sofifa_id] for sofifa_id in ids]
ordenar_jogadores_por_medias(jogadores)
tabela_jogadores.imprimir_jogadores(jogadores)
```
